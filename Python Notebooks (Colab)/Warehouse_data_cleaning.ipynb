```python
import pandas as pd
import numpy as np

# Make a working copy
df = df.copy()

# -----------------------------
# 1. Standardize column names
# -----------------------------
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(' ', '_')
)

# -----------------------------
# 2. Clean text columns
# -----------------------------
text_cols = [
    'location_code',
    'order_id',
    'sku',
    'bin',
    'zone',
    'order_tag',
    'store_id',
    'floor',
    'rack',
    'aisle',
    'pod_priority'
]

for col in text_cols:
    if col in df.columns:
        df[col] = df[col].astype('string').str.strip()

# -----------------------------
# 3. Convert dates
# -----------------------------
df['dt'] = pd.to_datetime(df['dt'], errors='coerce')
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')

# -----------------------------
# 4. Convert numeric columns
# -----------------------------
numeric_cols = [
    'order_qty',
    'bin_rank',
    'pods_per_picklist_in_that_zone',
    'length_in_cm',
    'width_in_cm',
    'height_in_cm',
    'weight_in_grams'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# -----------------------------
# 5. Remove exact duplicates
# -----------------------------
df = df.drop_duplicates().reset_index(drop=True)

print("Rows:", len(df))
print("Columns:", len(df.columns))
print("Duplicate rows:", df.duplicated().sum())
print("Missing values:")
print(df.isna().sum()[df.isna().sum() > 0])
```

```python
for col in ['floor', 'rack', 'aisle']:
    df[col] = df[col].fillna('UNKNOWN')
```

```python
print("Missing values:")
print(df.isna().sum()[df.isna().sum() > 0])
```

```python
df.to_csv('formatted.csv')
```

```python
df['weight_kg'] = df['weight_in_grams'] / 1000
```

```python
df['volume_cm3'] = (
    df['length_in_cm'] *
    df['width_in_cm'] *
    df['height_in_cm']
)
```

```python
df['priority_num'] = (
    df['pod_priority']
    .str.extract(r'(\d+)')[0]
    .astype('Int64')
)
```

```python
order_zone = (
    df.groupby(['order_id', 'zone'], as_index=False)
      .agg(
          order_date=('order_date', 'min'),
          store_id=('store_id', 'nunique'),
          priority_num=('priority_num', 'min'),

          sku_lines=('sku', 'count'),
          unique_skus=('sku', 'nunique'),

          total_qty=('order_qty', 'sum'),

          unique_bins=('bin', 'nunique'),

          total_weight_kg=('weight_kg', 'sum'),
          total_volume_cm3=('volume_cm3', 'sum'),

          zone_complexity_tier=(
              'pods_per_picklist_in_that_zone',
              'first'
          )
      )
)

order_zone.head()
```

```python
order_zone['store_count'] = order_zone['store_id']

order_zone['store_count'].value_counts().sort_index()
```

```python
order_zone['travel_seconds'] = (
    order_zone['unique_bins'] * 30
)

order_zone['pickup_seconds'] = (
    order_zone['total_qty'] * 5
)

order_zone['base_work_seconds'] = (
    order_zone['travel_seconds'] +
    order_zone['pickup_seconds']
)

order_zone['base_work_minutes'] = (
    order_zone['base_work_seconds'] / 60
)
```

```python
order_zone['picklists_by_qty'] = np.ceil(
    order_zone['total_qty'] / 2000
).astype(int)

order_zone['picklists_by_weight'] = np.ceil(
    order_zone['total_weight_kg'] / 200
).astype(int)

order_zone['picklists_required'] = (
    order_zone[['picklists_by_qty', 'picklists_by_weight']]
    .max(axis=1)
)
```

```python
order_zone['workload_level'] = pd.qcut(
    order_zone['base_work_minutes'],
    q=4,
    labels=['Low', 'Medium', 'High', 'Very High'],
    duplicates='drop'
)
```
